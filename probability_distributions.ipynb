```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial

# 1. Standard Normal Distribution
x = np.linspace(-4, 4, 200)
normal_y = (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * x**2)

# 2. Binomial Distribution (n=15, p=0.5)
k = np.arange(0, 16)
binomial_y = [factorial(15) / (factorial(int(i)) * factorial(15 - int(i))) * (0.5**15) for i in k]

# 3. Poisson Distribution (lambda=4)
poisson_y = [(4**int(i) * np.exp(-4)) / factorial(int(i)) for i in k]

fig, axs = plt.subplots(1, 3, figsize=(14, 4))
axs[0].plot(x, normal_y, color='blue', lw=2)
axs[0].set_title("Normal Distribution PDF")

axs[1].bar(k, binomial_y, color='green')
axs[1].set_title("Binomial PMF (n=15, p=0.5)")

axs[2].bar(k, poisson_y, color='darkorange')
axs[2].set_title("Poisson PMF (lambda=4)")

plt.tight_layout()
plt.savefig("distributions.png", dpi=300)
plt.show()
```
